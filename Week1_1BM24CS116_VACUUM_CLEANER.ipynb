```python
rooms = {
    "A": "Dirty",
    "B": "Dirty"
}

location = input("Enter starting location (A/B): ").upper()

def vacuum_agent(location, status):
    if status == "Dirty":
        return "Pick Dust"
    elif location == "A":
        return "Right"
    else:
        return "Left"

while rooms["A"] == "Dirty" or rooms["B"] == "Dirty":

    status = rooms[location]

    print("\nLocation:", location)
    print("Status:", status)

    action = vacuum_agent(location, status)

    print("Action:", action)

    if action == "Pick Dust":
        rooms[location] = "Clean"

    elif action == "Right":
        location = "B"

    elif action == "Left":
        location = "A"

print("\nBoth rooms are clean!")
```
