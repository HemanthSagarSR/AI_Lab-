```python
import math
import random

def calculate_conflicts(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j]:
                conflicts += 1
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_random_neighbor(state):
    neighbor = list(state)
    column = random.randint(0, 7)
    new_row = random.randint(0, 7)
    while new_row == state[column]:
        new_row = random.randint(0, 7)
    neighbor[column] = new_row
    return neighbor

def print_board(state):
    for row in range(8):
        line = ""
        for col in range(8):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

def simulated_annealing_8_queens(initial_state):
    current = list(initial_state)
    T = 10.0
    alpha = 0.995
    while T > 0.0001:
        current_cost = calculate_conflicts(current)
        if current_cost == 0:
            break
        nxt = get_random_neighbor(current)
        next_cost = calculate_conflicts(nxt)
        delta_E = next_cost - current_cost
        if delta_E < 0:
            current = nxt
        else:
            p = math.exp(-delta_E / T)
            if random.random() < p:
                current = nxt
        T *= alpha
    return current

if __name__ == "__main__":
    user_input = input("Enter initial state as 8 comma-separated numbers (e.g., 0,1,2,3,4,5,6,7): ")
    initial_state = [int(x.strip()) for x in user_input.split(",")]

    solution = simulated_annealing_8_queens(initial_state)
    final_conflicts = calculate_conflicts(solution)
    print(f"\nFinal State Array: {solution}")
    print(f"Remaining Conflicts: {final_conflicts}\n")
    print_board(solution)
```
