```python
def dls(state, goal, depth, path, visited):
    if state == goal:
        return list(path)

    if depth == 0:
        return None

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = []

    if row > 0:
        moves.append(zero - 3)
    if row < 2:
        moves.append(zero + 3)
    if col > 0:
        moves.append(zero - 1)
    if col < 2:
        moves.append(zero + 1)

    state_list = list(state)
    for move in moves:
        state_list[zero], state_list[move] = state_list[move], state_list[zero]
        new_state = tuple(state_list)

        if new_state not in visited:
            path.append(new_state)
            visited.add(new_state)

            result = dls(new_state, goal, depth - 1, path, visited)

            if result:
                return result

            visited.remove(new_state)
            path.pop()

        state_list[zero], state_list[move] = state_list[move], state_list[zero]

    return None


def ids(start, goal):

    depth = 0

    while True:
        path = [start]
        visited = {start}
        result = dls(start, goal, depth, path, visited)

        if result:
            return result

        depth += 1


print("Enter the puzzle values row by row")
print("Use 0 for blank space")

start = tuple(map(int, input().split()))
start += tuple(map(int, input().split()))
start += tuple(map(int, input().split()))

goal = (1, 2, 3 ,4 , 5 , 6 , 7, 8 , 0 )

solution = ids(start, goal)

print("\nSolution:")

for state in solution:
    print(state[0:3])
    print(state[3:6])
    print(state[6:9])
    print()
```
