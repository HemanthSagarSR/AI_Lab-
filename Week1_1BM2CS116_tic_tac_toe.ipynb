```python

board = [" "] * 9
print("Hemanth Sagar S R USN:1BM24CS116")

def show():
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])

def win(p):
    lines = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]

    for a,b,c in lines:
        if board[a] == board[b] == board[c] == p:
            return True
    return False

def minimax(turn):
    if win("O"):
        return 1
    if win("X"):
        return -1
    if " " not in board:
        return 0

    if turn:
        best = -10
        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                score = minimax(False)
                board[i] = " "
                best = max(best, score)
        return best

    best = 10
    for i in range(9):
        if board[i] == " ":
            board[i] = "X"
            score = minimax(True)
            board[i] = " "
            best = min(best, score)
    return best

def computer():
    best = -10
    move = 0

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            score = minimax(False)
            board[i] = " "

            if score > best:
                best = score
                move = i

    board[move] = "O"

while True:
    show()

    n = int(input("Enter position (1-9): ")) - 1

    if n < 0 or n > 8 or board[n] != " ":
        print("Invalid move")
        continue

    board[n] = "X"

    if win("X"):
        show()
        print("You win")
        break

    if " " not in board:
        show()
        print("Draw")
        break

    computer()

    if win("O"):
        show()
        print("Computer wins")
        break

    if " " not in board:
        show()
        print("Draw")
        break
```
