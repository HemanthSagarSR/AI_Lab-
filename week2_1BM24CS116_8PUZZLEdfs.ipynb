```python
def dfs(state, goal, visited, path, depth, max_depth):
    if state == goal:
        return path

    if depth >= max_depth:
        return None

    visited.add(state)

    zero = state.index(0)
    row = zero // 3
    col = zero % 3
    moves = []

    if row > 0: moves.append(zero - 3)  # Up
    if row < 2: moves.append(zero + 3)  # Down
    if col > 0: moves.append(zero - 1)  # Left
    if col < 2: moves.append(zero + 1)  # Right

    for move in moves:
        new_state = list(state)
        new_state[zero], new_state[move] = new_state[move], new_state[zero]
        new_state = tuple(new_state)

        if new_state not in visited:
            result = dfs(new_state, goal, visited, path + [new_state], depth + 1, max_depth)
            if result:
                return result

    visited.remove(state)
    return None

print("Enter the puzzle values row by row")
print("Use 0 for blank space")
start = tuple(map(int, input().split()))
start += tuple(map(int, input().split()))
start += tuple(map(int, input().split()))


goal = (1, 2, 3, 8, 0, 4, 7, 6, 5)
MAX_DEPTH = 5

solution = dfs(start, goal, set(), [start], 0, MAX_DEPTH)

if solution:
    print("\nfound soln")
    print(f"Moves taken: {len(solution) - 1}\n")
    for state in solution:
        print(state[0:3])
        print(state[3:6])
        print(state[6:9])
        print()
else:
    print("\nno soln")
```
